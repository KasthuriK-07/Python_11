```python
#If and Else condition
#Syntax:if expression(condition):
#Indentation Block 
#Eligibility to vote
age=int(input("Enter age to check eligibile for vote: "))
if age>=18:
    #Indentation Space
    print("Eligible for vote: ",age)
else:
    print("Not Eligible for vote: ",age)
```

```python
#2.Positive number-Square the data,Negative number-Cube the data
num=int(input("Enter number to check positive or negative: "))
if num>0:
    sqr=num**2
    print("Positive Number: +",num)
    print("Square value of positive number: ",sqr)
else:
    cube=num**3
    print("Negative number: ",num)
    print("Cube value of negative number: ",cube)
```

```python
#3.Check number is even or odd
n=int(input("Enter number to check even or odd: "))
if n%2!=0:
    print("Odd number: ",n)
else:
    print("Even number: ",n)
```

```python
#Check year is leap year or not
year=int(input("Enter year to check leap year or not:"))
if year%4==0:
    print("Leap Year: ",year)
else:
    print("Not leap Year: ",year)
```

```python
#Divisible by 7 and 5
n=int(input("Enter number to check divisible by both 7 and 5: "))
if n%7==0 and n%5==0:
    print("Divisible by both 7 and 5: ",n)
else:
    print("Not divisible by common 7 and 5: ",n)
```

```python
#Divisible by 7 and 5 and 2
n=int(input("Enter number to check divisible by both 7 and 5 and 2: "))
if n%7==0 and n%5==0 and n%2==0:
    print("Divisible by both 7 and 5 and 2: ",n)
else:
    print("Not divisible by common 7 and 5 and 2: ",n)
```

```python
market_price=float(input("Enter price: "))
if market_price>2500 and market_price<=3500:
    #10% discount
    discount_price=market_price*0.10
    total_bill=market_price-discount_price
elif market_price>3500 and market_price<=4500:
    discount_price=market_price*0.20
    total_bill=market_price-discount_price
elif market_price>4500 and market_price<=6500:
    discount_price=market_price*0.30
    total_bill=market_price-discount_price
else:
    pass#Empty statement
print("Original price: ",market_price)
print("Discount: ",discount_price)
print("Total Bill: ",total_bill)
```

```python
#Login Data 
#Nested-if
username=input("Enter username: ")
password=input("Enter password: ")
if username=="admin":
    if password=="admin":
        print("Login Successful")
    else:
        print("Password is not matching")
else:
    print("Invalid Credentials")
```
