```python
#Non Argument Function and Keyword Argument Function
#*args and **kwargs
def total(*n):
    sum=0
    for i in n:
        sum=sum+i
    return sum
```

```python
total(10,20)
```

```python
total(10,20,30,40)
```

```python
#Decorator-it is the function which adds extra functionality to the code without modifying original code
def my_decorator(func):
    def wrapper():
        print("Before function execution")
        func()
        print("After function execution")
    return wrapper

@my_decorator
def hello():
    print("Hello World")
```

```python
hello()
```

```python
import time
def timer(func):
    def wrapper():
        start=time.time()
        func()
        end=time.time()
        print("Total Execution Time: ",end-start)
    return wrapper

@timer
def login():
    print("Execution done")
login()
```

```python
#Generate-Produce the value at a time using yield
def count():
    for i in range(1,4):
        yield i
```

```python
for x in count():
    print(x)
```

```python
#Movie ticket booking system
def movie_ticket():
    print("1.Movies-1")
    print("2.Movies-2")
    movie=int(input("Enter the movie no to book the ticket"))
    print("Select the Screen")
    print("1.Screen-1")
    print("2.Screen-2")
    screen=int(input("Enter the screen no to book"))
    print("Timing Info")
    print("1.Morning")
    print("2.Aft")
    print("3.Evng")
    timing=int(input("Enter Timing: "))
    person=int(input("Enter the no of person: "))
    ticket_price=person*200
    print("Booking Details")
    print("Movie Name: ",movie)
    print("Screen Time: ",screen)
    print("Timings:",timing)
    print("No of Person: ",person)
    print("Total Ticket Price: ",ticket_price)
    print("Movie Booked Successfully")
```

```python
movie_ticket()
```
