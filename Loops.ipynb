```python
#Prime number
n=int(input("Enter number to check Prime number or not: "))
for i in range(2,n):
    if n%i==0:
        print("Not a Prime Number: ",n)
        print(n,"is divisible by: ",i)
        break #Stop the program
else:
    print("Prime Number: ",n)
```

```python
#345=543
#Reverse the number
n=345
sum=0
while n!=0:
    reminder=n%10
    sum=sum*10+reminder
    n=n//10
print(sum)
```

```python
#Reverse the number
n=307
temp=n
order=len(str(n))
sum=0
while n!=0:
    reminder=n%10
    sum=sum+reminder**order
    n=n//10
if temp==sum:
    print("Armstrong Number: ",sum)
else:
    print("Not Armstrong Number: ",sum)
```
